```python
import sympy as sp
sp.init_printing()
x = sp.Symbol('x')

expr = sp.exp(x**2-2) #exponent
expr2 = sp.diff(x+x-2) #derivative
expr_diff = expr.diff(x) #derivative

print(f"{sp.latex(expr)}")
print(f"{sp.latex(expr2)}")
print(f"{(expr_diff)}")
```

```python
print(f"{(expr_diff)}")

expr_diff_0 = expr.subs(x,0) #replace x with 0
print(f"{expr_diff_0}")

print(f"{ expr_diff_0.evalf()}") #evaluate the floating point value
```
